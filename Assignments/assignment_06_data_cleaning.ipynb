## 5. Convert and Clean Data Types

### Business framing:  

Sometimes columns that look like numbers are actually stored as text — which breaks calculations and slows down analysis. Common examples are price columns with dollar signs or availability stored as strings.

### Do the following:
- Identify one column with the wrong data type
- Clean and convert it into a usable format (e.g., from string to number)
- Check your work by summarizing or plotting the cleaned column

````python
import pandas as pd
import numpy as np
import re # Make sure regex module is imported for clean_bathrooms function

# Reload df to ensure it's available and fresh, in case original load was lost or kernel restarted
df = pd.read_csv("/content/listings.csv.gz")

# --- Re-applying previous cleaning steps to ensure df state is correct ---

# From GvQDqSfPSD-3: Drop columns with 100% missing values
columns_to_drop = [
    'neighborhood_overview', 'host_since', 'host_verifications', 'calendar_updated'
]
existing_columns_to_drop = [col for col in columns_to_drop if col in df.columns]
if existing_columns_to_drop:
    df = df.drop(columns=existing_columns_to_drop)

# From AfoW3qQeSJuB: Convert 'host_identity_verified' and 'host_has_profile_pic' to boolean
if 'host_identity_verified' in df.columns:
    df['host_identity_verified'] = df['host_identity_verified'].map({'t': True, 'f': False}).astype(bool)
if 'host_has_profile_pic' in df.columns:
    df['host_has_profile_pic'] = df['host_has_profile_pic'].map({'t': True, 'f': False}).astype(bool)

# From AfoW3qQeSJuB: Clean and convert 'bathrooms_text' to a numeric type
def clean_bathrooms(text):
    if pd.isna(text):
        return np.nan
    text = str(text).lower()
    if 'half-bath' in text:
        return 0.5
    match = re.search(r'(\d+\.?\d*)', text)
    if match:
        return float(match.group(1))
    return np.nan

if 'bathrooms_text' in df.columns:
    df['bathrooms'] = df['bathrooms_text'].apply(clean_bathrooms)
    # Check if the new 'bathrooms' column is substantially populated before dropping the original
    if df['bathrooms'].isnull().sum() < df.shape[0] / 2:
        df = df.drop(columns=['bathrooms_text'])

# --- Original code for y637bPKuSNOL: Cleaning 'price' column ---
# Identify 'price' as a column with an incorrect data type (object/string with currency symbols)

# Clean and convert 'price' column to numeric
# First, ensure it's a string, then remove '$' and ',' characters
if 'price' in df.columns and df['price'].dtype == 'object':
    print("Cleaning 'price' column...")
    df['price'] = df['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
    # Convert to numeric, coercing errors will turn invalid parsing into NaN
    df['price'] = pd.to_numeric(df['price'], errors='coerce')
    print("Finished cleaning 'price' column.")
else:
    print("'price' column not found or already numeric.")

# Check your work by summarizing the cleaned column
print("\nSummary of cleaned 'price' column:")
display(df['price'].describe())
````

If a listing appears twice, it could inflate revenue estimates or confuse users. Airbnb needs each listing to be unique and accurate.

### Do the following:
- Check for rows that are exact duplicates
- If your data has an ID column and each ID is supposed to unique, then make sure there are no duplicate IDs
- Remove duplicates if found

````python
import pandas as pd

# The DataFrame 'df' is already loaded and cleaned from previous steps.
# We will operate on the existing 'df' in memory.

print(f"Initial number of rows: {len(df)}")

# Check for exact duplicate rows across all columns
duplicate_rows = df.duplicated().sum()
if duplicate_rows > 0:
    print(f"\nFound {duplicate_rows} exact duplicate rows across all columns.")
else:
    print("\nNo exact duplicate rows found across all columns.")

# Check for duplicate 'id' values
# Assuming 'id' is supposed to be a unique identifier for each listing
duplicate_ids = df[df.duplicated(subset=['id'])]

if not duplicate_ids.empty:
    print(f"\nFound {len(duplicate_ids)} duplicate 'id' values. Examples:\n{duplicate_ids[['id']].head()}")

    # Remove duplicates based on the 'id' column, keeping the first occurrence
    df_cleaned = df.drop_duplicates(subset=['id'], keep='first')
    print(f"Removed {len(df) - len(df_cleaned)} rows with duplicate 'id' values.")
    df = df_cleaned.copy() # Update the original DataFrame
else:
    print("\nNo duplicate 'id' values found.")

print(f"Final number of rows after removing ID duplicates: {len(df)}")
````
